```python
# ============================================================
# KHO DỮ LIỆU TIẾNG VIỆT
# ============================================================

# Kỹ năng kỹ thuật theo ngành
TECH_SKILLS = {
    "it_backend": [
        "Python","Django","Flask","FastAPI","Node.js","Express","Java","Spring Boot",
        "PHP","Laravel","MySQL","PostgreSQL","MongoDB","Redis","Docker","Kubernetes",
        "REST API","GraphQL","Microservices","AWS","Git","Linux","CI/CD"
    ],
    "it_frontend": [
        "HTML","CSS","JavaScript","TypeScript","React","Vue.js","Angular","Next.js",
        "Tailwind CSS","Webpack","Vite","jQuery","Bootstrap","Figma","UI/UX","Responsive Design"
    ],
    "it_data": [
        "Python","R","TensorFlow","PyTorch","scikit-learn","Pandas","NumPy","Matplotlib",
        "SQL","Tableau","Power BI","Machine Learning","Deep Learning","NLP","Computer Vision","Spark"
    ],
    "it_devops": [
        "Docker","Kubernetes","Jenkins","GitLab CI","Terraform","Ansible","AWS","GCP","Azure",
        "Linux","Bash","Monitoring","Prometheus","Grafana","Nginx","Load Balancing"
    ],
    "marketing": [
        "Google Ads","Facebook Ads","SEO","Content Marketing","Email Marketing","Copywriting",
        "Photoshop","Canva","Google Analytics","Social Media","TikTok Ads","Brand Strategy",
        "Market Research","CRM","HubSpot"
    ],
    "accounting": [
        "Excel nâng cao","SAP","MISA","Kế toán thuế","Báo cáo tài chính","Kiểm toán",
        "Kế toán tổng hợp","Phân tích tài chính","Quyết toán","Thuế GTGT","Thuế TNDN"
    ],
    "sales": [
        "Kỹ năng bán hàng","Chăm sóc khách hàng","Đàm phán","CRM","Telesales",
        "B2B","B2C","Tìm kiếm khách hàng","Upselling","KPI doanh số","Thuyết trình"
    ],
    "hr": [
        "Tuyển dụng","Onboarding","C&B","HRIS","Đào tạo","BHXH","Hợp đồng lao động",
        "Đánh giá hiệu suất KPI","Văn hóa doanh nghiệp","Labor Law","Payroll"
    ]
}

# Kỹ năng mềm
SOFT_SKILLS = [
    "giao tiếp tốt","làm việc nhóm","tư duy logic","chịu áp lực cao",
    "quản lý thời gian","sáng tạo","cẩn thận","chi tiết","chủ động",
    "học hỏi nhanh","tiếng Anh tốt","lãnh đạo","phân tích vấn đề"
]

# Cấp độ kinh nghiệm
EXP_LEVELS = {
    "fresher": ["sinh viên năm cuối","mới ra trường","0 năm kinh nghiệm","thực tập","fresher","tốt nghiệp đại học"],
    "junior": ["1 năm kinh nghiệm","2 năm kinh nghiệm","junior","dưới 2 năm","có kinh nghiệm cơ bản"],
    "middle": ["3 năm kinh nghiệm","4 năm kinh nghiệm","middle","từ 3-5 năm","có kinh nghiệm thực tế"],
    "senior": ["5 năm kinh nghiệm","senior","trên 5 năm","nhiều năm kinh nghiệm","chuyên gia","lead"]
}

# Gợi ý vị trí theo kỹ năng
TITLE_MAPPING = {
    "it_backend": ["Backend Developer","Software Engineer","Full-stack Developer","API Developer","Java Developer","Python Developer","Node.js Developer"],
    "it_frontend": ["Frontend Developer","UI Developer","React Developer","Web Designer","Frontend Engineer"],
    "it_data": ["Data Scientist","Data Analyst","ML Engineer","AI Engineer","Data Engineer"],
    "it_devops": ["DevOps Engineer","SRE","Cloud Engineer","Infrastructure Engineer","System Administrator"],
    "marketing": ["Marketing Executive","Digital Marketing","Content Creator","SEO Specialist","Brand Manager"],
    "accounting": ["Kế toán viên","Kế toán tổng hợp","Kiểm toán viên","Kế toán trưởng","Chuyên viên tài chính"],
    "sales": ["Nhân viên kinh doanh","Sales Executive","Account Manager","Business Development","Key Account"],
    "hr": ["HR Executive","Chuyên viên tuyển dụng","HR Manager","C&B Specialist","Training & Development"]
}

print("✅ Đã định nghĩa kho dữ liệu tiếng Việt")
print(f"   - {sum(len(v) for v in TECH_SKILLS.values())} kỹ năng kỹ thuật")
print(f"   - {len(SOFT_SKILLS)} kỹ năng mềm")
print(f"   - {sum(len(v) for v in TITLE_MAPPING.values())} vị trí công việc")
```

```python
import random
random.seed(42)
np.random.seed(42)

def generate_cv(domain, exp_level):
    """Sinh text CV giả lập theo ngành và cấp độ"""
    skills = TECH_SKILLS[domain]
    n_skills = {"fresher": 3, "junior": 5, "middle": 8, "senior": 12}[exp_level]
    chosen_skills = random.sample(skills, min(n_skills, len(skills)))
    soft = random.sample(SOFT_SKILLS, 3)
    exp_phrase = random.choice(EXP_LEVELS[exp_level])
    edu = random.choice(["Đại học Bách Khoa","Đại học FPT","Đại học CNTT","Đại học Kinh tế","Cao đẳng nghề"])
    
    templates = [
        f"Tôi là {exp_phrase}, tốt nghiệp {edu}. Có kinh nghiệm với {', '.join(chosen_skills[:4])}. "
        f"Kỹ năng bổ sung: {', '.join(chosen_skills[4:])}. Kỹ năng mềm: {', '.join(soft)}.",
        
        f"Ứng viên {exp_phrase} chuyên về {', '.join(chosen_skills[:3])}. "
        f"Đã làm việc với {', '.join(chosen_skills[3:6])}. "
        f"Tốt nghiệp {edu}. Tính cách: {', '.join(soft)}.",
        
        f"Kinh nghiệm: {exp_phrase}. Kỹ năng chính: {', '.join(chosen_skills)}. "
        f"Học vấn: {edu}. Phẩm chất: {', '.join(soft)}."
    ]
    return random.choice(templates)

def generate_job(domain, exp_required):
    """Sinh text tin tuyển dụng theo ngành và yêu cầu kinh nghiệm"""
    skills = TECH_SKILLS[domain]
    n_req = {"fresher": 2, "junior": 4, "middle": 6, "senior": 9}[exp_required]
    required = random.sample(skills, min(n_req, len(skills)))
    preferred = random.sample([s for s in skills if s not in required], min(3, len(skills) - n_req))
    title = random.choice(TITLE_MAPPING[domain])
    exp_phrase = random.choice(EXP_LEVELS[exp_required])
    
    templates = [
        f"Tuyển dụng {title}. Yêu cầu: {', '.join(required)}. "
        f"Ưu tiên: {', '.join(preferred)}. Kinh nghiệm: {exp_phrase}.",
        
        f"Vị trí {title} - Cần người biết {', '.join(required[:3])} và có kinh nghiệm {exp_phrase}. "
        f"Thêm điểm nếu có: {', '.join(preferred)}.",
        
        f"Chúng tôi tìm kiếm {title} ({exp_phrase}). "
        f"Kỹ năng bắt buộc: {', '.join(required)}. Kỹ năng mong muốn: {', '.join(preferred)}."
    ]
    return random.choice(templates)

def compute_label(cv_domain, cv_exp, job_domain, job_exp):
    """Tính nhãn similarity dựa trên sự phù hợp ngành và kinh nghiệm"""
    exp_order = ["fresher", "junior", "middle", "senior"]
    
    if cv_domain == job_domain:
        base_score = 0.85
    elif cv_domain.startswith("it_") and job_domain.startswith("it_"):
        base_score = 0.50  # Cùng khối IT nhưng khác chuyên ngành
    else:
        base_score = 0.10  # Khác ngành hoàn toàn
    
    cv_lvl = exp_order.index(cv_exp)
    job_lvl = exp_order.index(job_exp)
    exp_diff = abs(cv_lvl - job_lvl)
    exp_penalty = [0, 0.1, 0.25, 0.45][exp_diff]
    
    score = base_score - exp_penalty + random.uniform(-0.05, 0.08)
    return float(np.clip(score, 0.0, 1.0))

# Sinh 800 cặp training data
print("Đang sinh dữ liệu huấn luyện...")
domains = list(TECH_SKILLS.keys())
exp_levels_list = ["fresher", "junior", "middle", "senior"]

train_data = []
for _ in range(800):
    cv_domain = random.choice(domains)
    cv_exp = random.choice(exp_levels_list)
    job_domain = random.choice(domains)
    job_exp = random.choice(exp_levels_list)
    
    cv_text = generate_cv(cv_domain, cv_exp)
    job_text = generate_job(job_domain, job_exp)
    label = compute_label(cv_domain, cv_exp, job_domain, job_exp)
    
    train_data.append({"cv": cv_text, "job": job_text, "label": label, 
                        "cv_domain": cv_domain, "cv_exp": cv_exp})

df = pd.DataFrame(train_data)
print(f"✅ Đã sinh {len(df)} cặp dữ liệu")
print(f"   Label trung bình: {df['label'].mean():.3f}")
print(f"   Phân phối ngành CV:")
print(df['cv_domain'].value_counts().to_string())
df.head(3)
```

## Phần 3 — Train mô hình CV-Job Matching

Dùng **Sentence-Transformers** với mô hình đa ngôn ngữ `paraphrase-multilingual-mpnet-base-v2`.

**Nguyên lý:**
- Encode CV và Job thành vector (embedding)
- Tính cosine similarity giữa 2 vector → điểm phù hợp 0–100%
- Fine-tune để mô hình hiểu ngữ cảnh tuyển dụng tiếng Việt


```python
# Chuẩn bị InputExample cho sentence-transformers
train_examples = []
for _, row in df.iterrows():
    train_examples.append(InputExample(
        texts=[row["cv"], row["job"]],
        label=float(row["label"])
    ))

# Chia train/val
split_idx = int(len(train_examples) * 0.85)
train_set = train_examples[:split_idx]
val_set = train_examples[split_idx:]
print(f"Train: {len(train_set)} | Val: {len(val_set)}")
```

```python
# Load mô hình base đa ngôn ngữ
print("Đang tải mô hình base (lần đầu có thể mất 2-3 phút)...")
MODEL_NAME = "paraphrase-multilingual-mpnet-base-v2"
model = SentenceTransformer(MODEL_NAME)
print(f"✅ Đã tải mô hình: {MODEL_NAME}")
print(f"   Max sequence length: {model.max_seq_length}")
```

```python
# Fine-tune với CosineSimilarityLoss
train_dataloader = DataLoader(train_set, shuffle=True, batch_size=32)
train_loss = losses.CosineSimilarityLoss(model)

EPOCHS = 4
WARMUP_STEPS = int(len(train_dataloader) * EPOCHS * 0.1)

print(f"Bắt đầu training: {EPOCHS} epochs, {len(train_dataloader)} bước/epoch")
print(f"Warmup steps: {WARMUP_STEPS}")

model.fit(
    train_objectives=[(train_dataloader, train_loss)],
    epochs=EPOCHS,
    warmup_steps=WARMUP_STEPS,
    show_progress_bar=True,
    output_path="./temp_matching_model"
)
print("\n✅ Training hoàn tất!")
```

```python
# Đánh giá mô hình trên tập validation
print("Đánh giá mô hình trên tập validation...")
val_cv_texts = [ex.texts[0] for ex in val_set]
val_job_texts = [ex.texts[1] for ex in val_set]
val_labels = np.array([ex.label for ex in val_set])

# Compute embeddings
cv_embeddings = model.encode(val_cv_texts, batch_size=32, show_progress_bar=False)
job_embeddings = model.encode(val_job_texts, batch_size=32, show_progress_bar=False)

# Compute cosine similarities
similarities = np.array([
    util.cos_sim(cv_emb, job_emb).item()
    for cv_emb, job_emb in zip(cv_embeddings, job_embeddings)
])

from scipy.stats import pearsonr, spearmanr
pearson_corr, _ = pearsonr(similarities, val_labels)
spearman_corr, _ = spearmanr(similarities, val_labels)

print(f"\n📊 Kết quả đánh giá:")
print(f"   Pearson correlation:  {pearson_corr:.4f}")
print(f"   Spearman correlation: {spearman_corr:.4f}")
print(f"   MAE (trung bình sai số): {np.mean(np.abs(similarities - val_labels)):.4f}")

# Demo thử nghiệm
print("\n🔍 Demo thử nghiệm:")
test_cases = [
    ("Tôi có 3 năm kinh nghiệm Python, Django, REST API, MySQL. Thành thạo Docker.",
     "Tuyển Backend Developer Python 3+ năm. Yêu cầu Django, REST API, Docker."),
    ("Sinh viên năm 4 CNTT, biết HTML CSS JavaScript cơ bản, có làm project React.",
     "Tuyển Kế toán viên có kinh nghiệm MISA, Excel, quyết toán thuế TNDN."),
    ("2 năm kinh nghiệm Marketing, Google Ads, Facebook Ads, SEO, Content.",
     "Tìm Digital Marketing Executive biết Google Ads và Facebook Ads, có portfolio."),
]
for cv, job in test_cases:
    cv_emb = model.encode(cv)
    job_emb = model.encode(job)
    score = util.cos_sim(cv_emb, job_emb).item()
    label = "✅ Phù hợp" if score > 0.6 else ("⚠️ Có thể" if score > 0.35 else "❌ Không phù hợp")
    print(f"   Score: {score:.3f} ({score*100:.1f}%) → {label}")
```

## Phần 4 — CV Analyzer (Phân tích hồ sơ)

Phân tích text CV và trả về:
- **Kỹ năng kỹ thuật** được trích xuất
- **Kỹ năng mềm** được phát hiện  
- **Cấp độ kinh nghiệm** (Fresher/Junior/Middle/Senior)
- **Vị trí gợi ý** phù hợp nhất
- **Tóm tắt ngắn** về ứng viên


```python
# ============================================================
# XÂY DỰNG CV ANALYZER
# ============================================================

# 1. Danh sách kỹ năng đầy đủ (tất cả ngành)
ALL_TECH_SKILLS = {}
for domain, skills in TECH_SKILLS.items():
    for skill in skills:
        ALL_TECH_SKILLS[skill.lower()] = {"skill": skill, "domain": domain}

# 2. Cấp độ kinh nghiệm - keywords
EXP_KEYWORDS = {
    "fresher": ["sinh viên","mới ra trường","fresher","thực tập","tốt nghiệp năm","0 năm","tìm môi trường"],
    "junior": ["1 năm","2 năm","junior","dưới 2","hơn 1 năm","1-2 năm","kinh nghiệm cơ bản"],
    "middle": ["3 năm","4 năm","middle","3-5 năm","vài năm","4-5 năm","kinh nghiệm thực tế"],
    "senior": ["5 năm","6 năm","7 năm","8 năm","9 năm","10 năm","senior","nhiều năm","chuyên gia",
               "trên 5","lead","tech lead","manager","trưởng"]
}

# 3. Sinh dữ liệu để train classifier cấp độ kinh nghiệm
X_exp, y_exp = [], []
for level, exp_list in EXP_LEVELS.items():
    for domain in domains:
        for _ in range(60):  # 60 mẫu mỗi combination
            cv = generate_cv(domain, level)
            X_exp.append(cv)
            y_exp.append(level)

X_train_exp, X_test_exp, y_train_exp, y_test_exp = train_test_split(
    X_exp, y_exp, test_size=0.2, random_state=42, stratify=y_exp
)

# TF-IDF + Logistic Regression
exp_vectorizer = TfidfVectorizer(ngram_range=(1,3), max_features=5000, analyzer='word')
X_train_vec = exp_vectorizer.fit_transform(X_train_exp)
X_test_vec = exp_vectorizer.transform(X_test_exp)

exp_classifier = LogisticRegression(max_iter=1000, C=1.0, random_state=42)
exp_classifier.fit(X_train_vec, y_train_exp)

print("📊 Kết quả phân loại cấp độ kinh nghiệm:")
y_pred = exp_classifier.predict(X_test_vec)
print(classification_report(y_test_exp, y_pred))
```

```python
def extract_skills_from_cv(cv_text):
    """Trích xuất kỹ năng kỹ thuật từ CV text"""
    cv_lower = cv_text.lower()
    found_skills = []
    found_domains = set()
    
    for skill_lower, info in ALL_TECH_SKILLS.items():
        # Tìm kỹ năng trong text (word boundary matching)
        pattern = r'\b' + re.escape(skill_lower) + r'\b'
        if re.search(pattern, cv_lower):
            found_skills.append(info["skill"])
            found_domains.add(info["domain"])
    
    return found_skills, list(found_domains)

def extract_soft_skills(cv_text):
    """Trích xuất kỹ năng mềm"""
    cv_lower = cv_text.lower()
    found = []
    for skill in SOFT_SKILLS:
        if skill.lower() in cv_lower:
            found.append(skill)
    return found

def predict_exp_level(cv_text):
    """Dự đoán cấp độ kinh nghiệm"""
    vec = exp_vectorizer.transform([cv_text])
    pred = exp_classifier.predict(vec)[0]
    proba = exp_classifier.predict_proba(vec)[0]
    classes = exp_classifier.classes_
    confidence = float(proba.max())
    return pred, confidence

def suggest_job_titles(domains_found):
    """Gợi ý vị trí công việc dựa trên ngành tìm được"""
    titles = []
    for domain in domains_found[:2]:  # Lấy tối đa 2 ngành chính
        if domain in TITLE_MAPPING:
            titles.extend(TITLE_MAPPING[domain][:3])
    return list(set(titles))[:5]

def generate_cv_summary(cv_text, tech_skills, exp_level, domains):
    """Tạo tóm tắt ngắn về ứng viên"""
    exp_label = {"fresher": "Fresher (mới ra trường)", "junior": "Junior (1-2 năm)",
                 "middle": "Middle (3-5 năm)", "senior": "Senior (5+ năm)"}
    domain_label = {
        "it_backend": "Lập trình Backend", "it_frontend": "Lập trình Frontend",
        "it_data": "Khoa học dữ liệu/AI", "it_devops": "DevOps/Hạ tầng",
        "marketing": "Marketing/Truyền thông", "accounting": "Kế toán/Tài chính",
        "sales": "Kinh doanh/Bán hàng", "hr": "Nhân sự/Hành chính"
    }
    
    main_domain = domains[0] if domains else "Đa ngành"
    domain_name = domain_label.get(main_domain, main_domain)
    
    top_skills = tech_skills[:5] if len(tech_skills) >= 5 else tech_skills
    skills_str = ", ".join(top_skills) if top_skills else "chưa xác định"
    
    return (f"Ứng viên cấp độ {exp_label.get(exp_level, exp_level)}, "
            f"chuyên ngành {domain_name}. "
            f"Kỹ năng nổi bật: {skills_str}. "
            f"Phát hiện {len(tech_skills)} kỹ năng kỹ thuật từ CV.")

def analyze_cv(cv_text):
    """Hàm phân tích CV tổng hợp"""
    tech_skills, domains = extract_skills_from_cv(cv_text)
    soft_skills = extract_soft_skills(cv_text)
    exp_level, exp_confidence = predict_exp_level(cv_text)
    suggested_titles = suggest_job_titles(domains)
    summary = generate_cv_summary(cv_text, tech_skills, exp_level, domains)
    
    return {
        "tech_skills": tech_skills,
        "soft_skills": soft_skills,
        "experience_level": exp_level,
        "exp_confidence": round(exp_confidence, 3),
        "primary_domains": domains[:3],
        "suggested_titles": suggested_titles,
        "summary": summary,
        "skill_count": len(tech_skills)
    }

# Test thử
test_cv = """Tôi là sinh viên năm 4 Đại học Bách Khoa, chuyên ngành CNTT.
Có 6 tháng thực tập tại công ty phần mềm, làm việc với Python, Django, MySQL và REST API.
Biết sử dụng Git, Docker cơ bản. Kỹ năng mềm: làm việc nhóm, giao tiếp tốt, học hỏi nhanh."""

result = analyze_cv(test_cv)
print("📄 Kết quả phân tích CV:")
for k, v in result.items():
    print(f"   {k}: {v}")
```

## Phần 5 — Career Chatbot (Tư vấn nghề nghiệp)

**Retrieval-based chatbot** — không cần train model nặng, dùng TF-IDF để tìm câu hỏi gần nhất trong database FAQ rồi trả lời.

Dataset gồm 60+ cặp Q&A về:
- Định hướng nghề nghiệp
- Kỹ năng cần thiết theo ngành

```python
# Xây dựng TF-IDF Retrieval Chatbot
class CareerChatbot:
    def __init__(self, faq_data):
        self.faq_data = faq_data
        self.questions = [item["question"].lower() for item in faq_data]
        self.answers = [item["answer"] for item in faq_data]
        
        # TF-IDF vectorizer với bigrams và trigrams
        self.vectorizer = TfidfVectorizer(
            ngram_range=(1, 3),
            analyzer="word",
            max_features=3000,
            sublinear_tf=True
        )
        self.question_vectors = self.vectorizer.fit_transform(self.questions)
        print(f"✅ Chatbot khởi tạo với {len(faq_data)} câu hỏi, {self.question_vectors.shape[1]} features")
    
    def get_answer(self, user_query, confidence_threshold=0.15):
        query_processed = user_query.lower().strip()
        query_vec = self.vectorizer.transform([query_processed])
        similarities = cosine_similarity(query_vec, self.question_vectors)[0]
        
        best_idx = int(np.argmax(similarities))
        best_score = float(similarities[best_idx])
        
        if best_score >= confidence_threshold:
            return {
                "answer": self.answers[best_idx],
                "matched_question": self.faq_data[best_idx]["question"],
                "confidence": round(best_score, 3),
                "found": True
            }
        else:
            return {
                "answer": ("Xin lỗi, tôi chưa có thông tin về câu hỏi này. "
                           "Bạn có thể hỏi về: định hướng nghề nghiệp, kỹ năng cần học, "
                           "mức lương thị trường, cách viết CV, hoặc tips phỏng vấn?"),
                "matched_question": None,
                "confidence": round(best_score, 3),
                "found": False
            }

chatbot = CareerChatbot(FAQ_DATA)

# Test chatbot
test_queries = [
    "Tôi muốn học lập trình web, bắt đầu từ đâu?",
    "Lương kỹ sư phần mềm mới ra trường là bao nhiêu?",
    "Nên học React hay Vue cho frontend?",
    "CV của tôi cần có những gì?",
    "Tôi muốn ăn phở ở đâu ngon?"  # Câu ngoài lĩnh vực
]

print("🤖 Test chatbot:")
for q in test_queries:
    result = chatbot.get_answer(q)
    status = "✅" if result["found"] else "❓"
    print(f"\n{status} Q: {q}")
    print(f"   Confidence: {result['confidence']:.3f}")
    print(f"   A: {result['answer'][:100]}...")
```
